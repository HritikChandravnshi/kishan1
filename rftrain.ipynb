```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
import pickle

# Load the dataset
df = pd.read_csv('crop_disease_environmental_data_with_weather.csv')

# Get unique plant names
plants = df['Disease'].apply(lambda x: x.split('___')[0]).unique()

# Initialize OneHotEncoder for the weather conditions
encoder = OneHotEncoder(sparse=False, drop='first')  # drop='first' to avoid dummy variable trap

# Train a separate Random Forest model for each plant type
for plant in plants:
    # Filter the dataset for the specific plant
    plant_df = df[df['Disease'].str.startswith(f'{plant}___')]
    
    # Check if plant_df is empty
    if plant_df.empty:
        print(f"No data available for plant: {plant}. Skipping model training.")
        continue
    
    # Prepare the features and target variable
    X = plant_df[['Temperature', 'Weather_Conditions', 'Humidity']]
    y = plant_df['Disease']
    
    # One-hot encode the Weather_Conditions column
    X_encoded = pd.DataFrame(encoder.fit_transform(X[['Weather_Conditions']]), columns=encoder.get_feature_names_out())
    X_encoded = pd.concat([X[['Temperature', 'Humidity']].reset_index(drop=True), X_encoded.reset_index(drop=True)], axis=1)
    
    # Check if there are enough samples to split the data
    if len(plant_df) < 2:  # Adjust this condition if necessary
        print(f"Not enough data to train the model for plant: {plant}. Skipping model training.")
        continue
    
    # Split the dataset into training and test sets
    X_train, X_test, y_train, y_test = train_test_split(X_encoded, y, test_size=0.2, random_state=42)
    
    # Initialize and train the Random Forest model
    rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
    rf_model.fit(X_train, y_train)
    
    # Save the trained model to a file specific to the plant
    model_filename = f'rf_model_{plant}.pkl'
    with open(model_filename, 'wb') as file:
        pickle.dump(rf_model, file)
    
    print(f"Model for {plant} saved as '{model_filename}'")
```
